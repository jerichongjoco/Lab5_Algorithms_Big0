```python
# Appendix 3: E-Commerce Order Processing. This notebook compares the performance of Bubble Sort, Selection Sort, and Insertion Sort under different initial order conditions.

import random

# Student-specific inputs
LAST_NAME = "Ongjoco"
STUDENT_ID = "TUPM-26-1852"
FAVORITE_ARTIST = "Cup of Joe"
SEED_DIGIT = int(STUDENT_ID[-1])  # 2
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
ARTIST_SUM = sum(ord(char.upper()) - ord("A") + 1 for char in FAVORITE_ARTIST if char.isalpha())

base_orders = [int(STUDENT_ID[-2:]) * 2, ID_SUM, ARTIST_SUM, SEED_DIGIT * 15, len(FAVORITE_ARTIST) * 10, 45, 12, 88, 33, 5]

sorted_orders = sorted(base_orders)
reverse_orders = sorted(base_orders, reverse=True)
random_orders = base_orders.copy()
random.seed(SEED_DIGIT)
random.shuffle(random_orders)

print("=== 3. E-COMMERCE ORDER PROCESSING ===")
print(f"Student: {LAST_NAME} ({STUDENT_ID})")
print(f"Sorted input:    {sorted_orders}")
print(f"Reverse input:   {reverse_orders}")
print(f"Random input:    {random_orders}")
```

```python
def bubble_sort(arr):
    data = arr.copy()
    comps = moves = 0
    n = len(data)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            comps += 1
            if data[j] > data[j + 1]:
                data[j], data[j + 1] = data[j + 1], data[j]
                moves += 1
                swapped = True
        if not swapped:
            break
    return data, comps, moves

def selection_sort(arr):
    data = arr.copy()
    comps = moves = 0
    n = len(data)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comps += 1
            if data[j] < data[min_idx]:
                min_idx = j
        if min_idx != i:
            data[i], data[min_idx] = data[min_idx], data[i]
            moves += 1
    return data, comps, moves

def insertion_sort(arr):
    data = arr.copy()
    comps = moves = 0
    for i in range(1, len(data)):
        key = data[i]
        j = i - 1
        while j >= 0:
            comps += 1
            if data[j] > key:
                data[j + 1] = data[j]
                moves += 1
                j -= 1
            else:
                break
        data[j + 1] = key
    return data, comps, moves

conditions = {
    "Sorted": sorted_orders,
    "Reverse-Ordered": reverse_orders,
    "Random": random_orders
}

for cond_name, dataset in conditions.items():
    print(f"\n--- Condition: {cond_name} ---")
    for name, sort_func in [("Bubble Sort", bubble_sort), ("Selection Sort", selection_sort), ("Insertion Sort", insertion_sort)]:
        res, comps, moves = sort_func(dataset)
        print(f"{name:15} | Comparisons: {comps:3} | Movements: {moves:3} | Correctly Sorted: {res == sorted(dataset)}")
```
