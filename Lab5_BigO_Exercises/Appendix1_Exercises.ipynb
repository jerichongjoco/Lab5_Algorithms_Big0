```python
# Appendix 1: Emergency Dispatch Record Search. This notebook develops and analyzes a solution for locating a target incident in a rotated sorted record list.
    
# Student-specific inputs
LAST_NAME = "Ongjoco"
STUDENT_ID = "TUPM-26-1852"
FAVORITE_ARTIST = "Cup of Joe"
SEED_DIGIT = int(STUDENT_ID[-1])  # 2
ID_SUM = sum(int(digit) for digit in STUDENT_ID if digit.isdigit())
ARTIST_SUM = sum(ord(char.upper()) - ord("A") + 1 for char in FAVORITE_ARTIST if char.isalpha())

# Data Generation Rules
# 1. Generate unique values for incident records
original_dataset = sorted([i * 3 + SEED_DIGIT * 5 for i in range(1, 16)])

# 2. Determine rotation point using SEED_DIGIT
rotation_index = SEED_DIGIT % len(original_dataset)
rotated_dataset = original_dataset[rotation_index:] + original_dataset[:rotation_index]

# 3. Generate target incident using ID_SUM and ARTIST_SUM
target_incident = original_dataset[(ID_SUM + ARTIST_SUM) % len(original_dataset)]

print("=== 1. EMERGENCY DISPATCH RECORD SEARCH ===")
print(f"Original Dataset: {original_dataset}")
print(f"Rotated Dataset:  {rotated_dataset}")
print(f"Target Incident:  {target_incident}")
```

```python
# Search in Rotated Sorted Array (Modified Binary Search)
def search_rotated_array(arr, target):
    left, right = 0, len(arr) - 1
    comparisons = 0
    
    while left <= right:
        comparisons += 1
        mid = (left + right) // 2
        
        if arr[mid] == target:
            return mid, comparisons
        
        # Check if left half is sorted
        if arr[left] <= arr[mid]:
            comparisons += 1
            if arr[left] <= target < arr[mid]:
                right = mid - 1
            else:
                left = mid + 1
        else: # Right half is sorted
            if arr[mid] < target <= arr[right]:
                left = mid + 1
            else:
                right = mid - 1
                
    return -1, comparisons

# Linear Search for comparison
def linear_search(arr, target):
    comparisons = 0
    for i, val in enumerate(arr):
        comparisons += 1
        if val == target:
            return i, comparisons
    return -1, comparisons

position, rotated_comps = search_rotated_array(rotated_dataset, target_incident)
linear_pos, linear_comps = linear_search(rotated_dataset, target_incident)

print("\n--- RESULTS ---")
print(f"Target Position in Rotated List: {position if position != -1 else 'Not Found'}")
print(f"Rotated Search Comparisons: {rotated_comps}")
print(f"Linear Search Comparisons: {linear_comps}")
print(f"Time Complexity (Rotated Search): O(log n)")
print(f"Space Complexity: O(1)")
```
