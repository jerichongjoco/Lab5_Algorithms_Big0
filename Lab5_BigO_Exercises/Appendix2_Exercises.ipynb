```python
# Appendix 2: University Enrollment Records. This notebook consolidates two sorted enrollment datasets from different departments into one ordered collection.

# Student-specific inputs
LAST_NAME = "Ongjoco"
STUDENT_ID = "TUPM-26-1852"
FAVORITE_ARTIST = "Cup of Joe"
SEED_DIGIT = int(STUDENT_ID[-1])  # 2
ARTIST_LENGTH = len([c for c in FAVORITE_ARTIST if c.isalpha()])
NAME_LENGTH = len(LAST_NAME)

dataset_a = sorted([i * 2 + SEED_DIGIT for i in range(1, 8)])
dataset_b = sorted([i * 3 + ARTIST_LENGTH for i in range(1, 7)])

print("=== 2. UNIVERSITY ENROLLMENT RECORDS ===")
print(f"Student: {LAST_NAME} ({STUDENT_ID})")
print(f"Department A Dataset: {dataset_a}")
print(f"Department B Dataset: {dataset_b}")
```

```python
# Consolidate two sorted lists (Merge process)
def merge_sorted_datasets(list_a, list_b):
    merged = []
    i = j = 0
    comparisons = 0
    
    while i < len(list_a) and j < len(list_b):
        comparisons += 1
        if list_a[i] <= list_b[j]:
            merged.append(list_a[i])
            i += 1
        else:
            merged.append(list_b[j])
            j += 1
            
    while i < len(list_a):
        merged.append(list_a[i])
        i += 1
        
    while j < len(list_b):
        merged.append(list_b[j])
        j += 1
        
    return merged, comparisons

consolidated_dataset, total_comparisons = merge_sorted_datasets(dataset_a, dataset_b)

print("\n--- CONSOLIDATION RESULTS ---")
print(f"Consolidated Dataset: {consolidated_dataset}")
print(f"Number of records before consolidation: Department A ({len(dataset_a)}), Department B ({len(dataset_b)})")
print(f"Number of records after consolidation: {len(consolidated_dataset)}")
print(f"Comparison Count: {total_comparisons}")
print(f"Ordering Verification: {'Sorted correctly' if consolidated_dataset == sorted(consolidated_dataset) else 'Error'}")
print(f"Time Complexity: O(n + m)")
print(f"Space Complexity: O(n + m)")
```
